#### Get csv and write as parquet


```python_glue_session
spark = SparkSession.builder.getOrCreate()
```

```python_glue_session
def csv_to_parquet(csv_path, parquet_path):
    
    df = spark.read.csv(
        csv_path,
        header=True,        
        quote='"',          
        escape='"',         
        multiLine=True,     
        inferSchema=True    
    )
    
    df.write.mode("overwrite").parquet(parquet_path)
```

```python_glue_session
from pyspark.sql import SparkSession
paths = [("s3://novaims-group30-bdf/ml32m/bronze/links","s3://novaims-group30-bdf/ml32m/silver/links"), 
         ("s3://novaims-group30-bdf/ml32m/bronze/movies","s3://novaims-group30-bdf/ml32m/silver/movies"),
         ("s3://novaims-group30-bdf/ml32m/bronze/ratings","s3://novaims-group30-bdf/ml32m/silver/ratings"),
         ("s3://novaims-group30-bdf/ml32m/bronze/tags","s3://novaims-group30-bdf/ml32m/silver/tags")]

for csv_path,parquet_path in paths:
    csv_to_parquet(csv_path, parquet_path)
```
